5. **Ділення з перевіркою типів.**  

  1. Напишіть функцію `int_division` яка приймає на вхід два аргументи типу `int` - `a` і `b`, і виконує ділення першого аргумента на другий: `a/b`. При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0. Якщо другий аргумент не 0, фукнція вертає результат ділення, інакше - замість виконання ділення, вертає повідомлення на екран "Не можу виконати ділення на 0".

    Якщо типи, надані у функцію, не `int` - функція виводить повідомлення "Не працюю з такими типами даних.".

  2. Запустіть функцію на наступних парах чисел:
- 12 і 6
- 20 і 0
- 112 і 10
- 99 і 0
- 11 i 1.1

    і виведіть на екран повідомлення "Результат роботи функції: " - і далі те, що повернула функція.

  3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

````python
division_pairs = [(12, 6), (20, 0), (112, 10), (99, 0), (11, 1.1)]

def int_division(a = int, b = int):
  if not isinstance(a, int) or not isinstance(b, int):
    print('Не працюю з такими типами даних.')
  elif b == 0:
    print('Не можу виконати ділення на "0".')
  else:
    return a/b

for a, b in division_pairs:
  print(f'Зараз ділимо пару {a, b}')
  print(f'Результат роботи функції: {int_division(a, b)}')
````

Функція повертає None у тих випадках, коли return залишається пустий. Оскільки результат умови для перевірки b = 0 лише виводить заданий текст на екран, а не зберігає його як результат функції, то фактично результату функції при діленні на 0 немає, тобто результат функції None. Тому при виведенні результату функції через друк повідомлення ми отримуємо і текст про неділення на 0, який пропсианий в умові функції, і повідомлення про None, який є власне результатом функції. Якщо ж лише викликати функцію, то ми так само отримаємо повідомлення про неділення на 0, але те, що результат функції None, ми просто не виводимо повідомленням (цю різницю видно нижче)

````python
def int_division(a = int, b = int):
  if not isinstance(a, int) or not isinstance(b, int):
    print('Не працюю з такими типами даних.')
  elif b == 0:
    print('Не можу виконати ділення на "0".')
  else:
    return a/b

int_division(20, 0)
print('\n')
print(int_division(20, 0))
````

````python
division_pairs = [(12, 6), (20, 0), (112, 10), (99, 0), (11, 1.1)]

def int_division(a = int, b = int):
  if not isinstance(a, int) or not isinstance(b, int):
    return 'Не працюю з такими типами даних.'
  elif b == 0:
    return 'Не можу виконати ділення на "0".'
  else:
    return a/b

for a, b in division_pairs:
  print(f'Зараз ділимо пару {a, b}')
  print(f'Результат роботи функції: {int_division(a, b)}')
````
